```python
# Question 3 

import os

# Create package folder
os.makedirs("studentpackage", exist_ok=True)

# Create __init__.py
with open("studentpackage/__init__.py", "w") as file:
    file.write("")

# Create marks.py
with open("studentpackage/marks.py", "w") as file:
    file.write("""
def total(a, b, c):
    return a + b + c

def average(a, b, c):
    return (a + b + c) / 3
""")

# Create result.py
with open("studentpackage/result.py", "w") as file:
    file.write("""
def grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 40:
        return "D"
    else:
        return "Fail"
""")

print("Package created successfully")

```

```python
# Question 3 Continued

from studentpackage.marks import total, average
from studentpackage.result import grade

m1 = 80
m2 = 75
m3 = 85

t = total(m1, m2, m3)
avg = average(m1, m2, m3)

print("Total Marks:", t)
print("Average:", avg)
print("Grade:", grade(avg))

```

```python
# Question 4 (a)

import re

email = input("Enter email address: ")

pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'

if re.fullmatch(pattern, email):
    print("Valid email address")
else:
    print("Invalid email address")
```

```python
# Question 4 (b)

import re

text = """
Contact Rahul at 9876543210.
Aman can be reached at 8765432109.
The office code is 12345.
Riya's number is 7012345678.
"""

pattern = r'\b[6-9]\d{9}\b'

numbers = re.findall(pattern, text)

print("Mobile numbers found:")

for number in numbers:
    print(number)
```

```python
# Question 4 (c)

import re

text = "Python 3 is easy. Student marks are 85 and 92."

# .  -> Any character except newline
print("Dot:", re.findall(r'P.thon', text))

# ^  -> Beginning of string
print("Starts with Python:", bool(re.search(r'^Python', text)))

# $  -> End of string
print("Ends with 92:", bool(re.search(r'92\.$', text)))

# \d -> Digit
print("Digits:", re.findall(r'\d', text))

# \d+ -> One or more digits
print("Numbers:", re.findall(r'\d+', text))

# \w+ -> Words
print("Words:", re.findall(r'\w+', text))

# \s -> Whitespace
print("Spaces:", re.findall(r'\s', text))

# [0-9] -> Character set
print("Numbers using []:", re.findall(r'[0-9]+', text))
```
